```python
def replace_byte(byteString: bytes, idx: int, newNumber:int ) -> bytes :
    myByteArray = bytearray(byteString)
    myByteArray[idx] = newNumber
    new_bytes = bytes(myByteArray)
    return new_bytes


iv= replace_byte(iv, 0, 0)
print(iv)
```
